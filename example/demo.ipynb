# Demo-Notebook für Jupyter Auto Scroll

Führe **"Run All"** aus (oben in der Toolbar oder per Befehlspalette `Notebook: Run All Cells`).
Jede Zelle unten braucht 1–3 Sekunden, damit du gut sehen kannst, wie die Ansicht automatisch zur jeweils laufenden Zelle scrollt.

```python
import time

def laufen(nummer, sekunden):
    print(f"Zelle {nummer} gestartet ({sekunden:.1f}s) ...")
    time.sleep(sekunden)
    print(f"Zelle {nummer} fertig.")
```

```python
laufen(1, 1.5)
```

```python
laufen(2, 2.0)
```
